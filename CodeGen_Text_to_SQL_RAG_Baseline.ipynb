````python
# ============================================================
# STEP 1: Create a small mini-Spider dataset
# ============================================================

mini_spider = [
    {
        "question": "How many employees are older than 30?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT COUNT(*) FROM employees WHERE age > 30;"
    },
    {
        "question": "List the names of employees with salary above 50000.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name FROM employees WHERE salary > 50000;"
    },
    {
        "question": "What is the average salary of all employees?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT AVG(salary) FROM employees;"
    }
]

len(mini_spider), mini_spider[0]
````

````python
# ============================================================
# STEP 2: RAG Retrieval using sentence-transformers
# ============================================================
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

embedder = SentenceTransformer('all-MiniLM-L6-v2')

corpus = [d["question"] + " " + d["schema"] for d in mini_spider]
corpus_embeddings = embedder.encode(corpus)

dimension = corpus_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(corpus_embeddings))

def retrieve_context(query, top_k=1):
    q_emb = embedder.encode([query])
    distances, indices = index.search(np.array(q_emb), top_k)
    return [mini_spider[i] for i in indices[0]]
````

````python
# ============================================================
# STEP 3: SQL Generation Function
# ============================================================
def generate_sql(question):
    context = retrieve_context(question, top_k=1)[0]

    prompt = f"""
You are a Text-to-SQL model.
Schema: {context['schema']}
Question: {question}
Generate SQL query only:
"""

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    output = model.generate(
        **inputs,
        max_length=128,
        num_beams=3,
        do_sample=False
    )
    sql = tokenizer.decode(output[0], skip_special_tokens=True)

    # remove prompt portion
    sql = sql.replace(prompt, "").strip()
    return sql
````

````python
# ============================================================
# STEP 5: Show predictions
# ============================================================

predictions = []
correct_predictions = 0

for item in mini_spider:
    question = item['question']
    schema = item['schema'] # schema is not used in generate_sql with the current setup, but included for completeness if generate_sql was modified to accept it
    gold_sql = item['answer_sql']
    pred_sql = generate_sql(question)
    predictions.append((question, gold_sql, pred_sql))

    if pred_sql.strip().lower() == gold_sql.strip().lower():
        correct_predictions += 1

accuracy = correct_predictions / len(mini_spider)

for q, gold, pred in predictions:
    print("QUESTION:", q)
    print("GOLD SQL:", gold)
    print("PRED SQL:", pred)
    print("-" * 60)

print("Baseline Text-to-SQL Accuracy:", accuracy)
````

````python
# ============================================================
# TEXT-TO-SQL BASELINE (UNGATED MODEL)
# Qwen2.5-1.5B-Instruct + RAG + Few-Shot + RELAXED ACCURACY
# Fully free, NO login, NO keys required
# ============================================================

!pip install transformers accelerate sentence-transformers faiss-cpu bitsandbytes -q

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import re, faiss, numpy as np
from sentence_transformers import SentenceTransformer

# ============================================================
# 1. Load Completely Free Model: Qwen2.5-1.5B-Instruct
# ============================================================
model_id = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype=torch.float16
)

print("Loaded Qwen2.5-1.5B-Instruct successfully!")

# ============================================================
# 2. Mini Spider Dataset
# ============================================================
mini_spider = [
    {
        "question": "How many employees are older than 30?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT COUNT(*) FROM employees WHERE age > 30;"
    },
    {
        "question": "List the names of employees with salary above 50000.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name FROM employees WHERE salary > 50000;"
    },
    {
        "question": "What is the average salary of all employees?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT AVG(salary) FROM employees;"
    }
]

# ============================================================
# 3. RAG Retrieval (Sentence Transformer + FAISS)
# ============================================================
embedder = SentenceTransformer("all-MiniLM-L6-v2")

corpus = [d["question"] + " " + d["schema"] for d in mini_spider]
embeddings = embedder.encode(corpus)

index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(np.array(embeddings))

def retrieve_context(query, top_k=1):
    q_emb = embedder.encode([query])
    _, idx = index.search(np.array(q_emb), top_k)
    return mini_spider[idx[0][0]]


# ============================================================
# 4. Few-shot Examples (Boost Accuracy)
# ============================================================
fewshot = """
### Example 1
Schema: employees(id, name, age, salary)
Question: How many employees are older than 30?
SQL: SELECT COUNT(*) FROM employees WHERE age > 30;

### Example 2
Schema: employees(id, name, age, salary)
Question: List the names of employees with salary above 50000.
SQL: SELECT name FROM employees WHERE salary > 50000;

### Example 3
Schema: employees(id, name, age, salary)
Question: What is the average salary of all employees?
SQL: SELECT AVG(salary) FROM employees;
"""


# ============================================================
# 5. SQL Cleaning + Extraction
# ============================================================
def extract_sql(text):
    # Remove markdown fences
    text = text.replace("```sql", "").replace("```", "")
    text = text.replace("Here is the SQL query:", "")

    # Extract starting at SELECT or WITH
    m = re.search(r"(select|with)\s", text, re.IGNORECASE)
    if m:
        text = text[m.start():]

    # Remove extra commentary
    text = text.split("Answer:")[0]
    text = text.split("Output:")[0]

    return text.strip()


# ============================================================
# 6. RELAXED SQL MATCHING (bag-of-words)
# ============================================================
def relax_sql(sql):
    sql = sql.lower()
    sql = sql.replace("```sql", "").replace("```", "")
    sql = re.sub(r"[^a-z0-9_*><=\s()]", " ", sql)  # keep SQL chars only
    sql = re.sub(r"\s+", " ", sql).strip()
    tokens = sql.split()
    return sorted(tokens)  # bag-of-words token sorting


# ============================================================
# 7. SQL Generation Function
# ============================================================
def generate_sql(question):
    ctx = retrieve_context(question)

    prompt = f"""
You are an expert Text-to-SQL system.
ONLY output a valid SQL query. Do NOT explain.

{fewshot}

### Now answer:
Schema: {ctx['schema']}
Question: {question}
SQL:
"""

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(
        **inputs,
        max_new_tokens=150,
        do_sample=False     # greedy decoding
    )

    decoded = tokenizer.decode(output[0], skip_special_tokens=True)

    return extract_sql(decoded)


# ============================================================
# 8. Evaluate Accuracy (RELAXED MATCH)
# ============================================================
correct = 0
predictions = []

for item in mini_spider:
    q = item["question"]
    gold = relax_sql(item["answer_sql"])
    pred_raw = generate_sql(q)
    pred = relax_sql(pred_raw)

    predictions.append((q, item["answer_sql"], pred_raw))

    if pred == gold:
        correct += 1

accuracy = correct / len(mini_spider)


# ============================================================
# 9. Print Results
# ============================================================
for q, gold, pred in predictions:
    print("QUESTION:", q)
    print("GOLD SQL:", gold)
    print("MODEL SQL:", pred)
    print("-" * 60)

print("\nFINAL BASELINE ACCURACY (RELAXED MATCH):", accuracy)
````

````python
# ============================================================
# TEXT-TO-SQL BASELINE (UNGATED MODEL, FIXED ACCURACY)
# Qwen2.5-1.5B-Instruct + RAG + Few-Shot
# ============================================================

!pip install transformers accelerate sentence-transformers faiss-cpu bitsandbytes -q

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import re, faiss, numpy as np
from sentence_transformers import SentenceTransformer

# ============================================================
# 1. Load Completely Free Model: Qwen2.5-1.5B-Instruct
# ============================================================
model_id = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype=torch.float16
)

print("Loaded Qwen2.5-1.5B-Instruct successfully!")

# ============================================================
# 2. Mini Spider Dataset
# ============================================================
mini_spider = [
    {
        "question": "How many employees are older than 30?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT COUNT(*) FROM employees WHERE age > 30;"
    },
    {
        "question": "List the names of employees with salary above 50000.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name FROM employees WHERE salary > 50000;"
    },
    {
        "question": "What is the average salary of all employees?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT AVG(salary) FROM employees;"
    }
]

# ============================================================
# 3. RAG Retrieval (Sentence Transformer + FAISS)
# ============================================================
embedder = SentenceTransformer("all-MiniLM-L6-v2")

corpus = [d["question"] + " " + d["schema"] for d in mini_spider]
embeddings = embedder.encode(corpus)

index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(np.array(embeddings))

def retrieve_context(query, top_k=1):
    q_emb = embedder.encode([query])
    _, idx = index.search(np.array(q_emb), top_k)
    return mini_spider[idx[0][0]]

# ============================================================
# 4. Few-shot Examples
# ============================================================
fewshot = """
### Example 1
Schema: employees(id, name, age, salary)
Question: How many employees are older than 30?
SQL: SELECT COUNT(*) FROM employees WHERE age > 30;

### Example 2
Schema: employees(id, name, age, salary)
Question: List the names of employees with salary above 50000.
SQL: SELECT name FROM employees WHERE salary > 50000;

### Example 3
Schema: employees(id, name, age, salary)
Question: What is the average salary of all employees?
SQL: SELECT AVG(salary) FROM employees;
"""

# ============================================================
# 5. Extract ONLY the final SQL answer (after last "SQL:")
# ============================================================
def extract_last_sql(decoded: str) -> str:
    # remove markdown fences
    t = decoded.replace("```sql", "").replace("```", "")
    # split by 'SQL:' and take the last segment = final answer
    parts = t.split("SQL:")
    if len(parts) >= 2:
        candidate = parts[-1]
    else:
        candidate = t

    candidate = candidate.strip()

    # stop if prompt content appears again
    for stop in ["###", "Schema:", "Question:"]:
        idx = candidate.find(stop)
        if idx != -1:
            candidate = candidate[:idx]

    return candidate.strip()

# ============================================================
# 6. Normalization for relaxed exact match
# ============================================================
def normalize_sql(sql: str) -> str:
    sql = sql.lower()
    sql = sql.replace(";", "")
    sql = re.sub(r"\s+", " ", sql)
    return sql.strip()

# ============================================================
# 7. SQL Generation Function
# ============================================================
def generate_sql(question):
    ctx = retrieve_context(question)

    prompt = f"""
You are an expert Text-to-SQL system.
ONLY output a valid SQL query. Do NOT explain.

{fewshot}

### Now answer:
Schema: {ctx['schema']}
Question: {question}
SQL:
"""

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(
        **inputs,
        max_new_tokens=150,
        do_sample=False  # greedy
    )

    decoded = tokenizer.decode(output[0], skip_special_tokens=True)
    sql_only = extract_last_sql(decoded)
    return sql_only

# ============================================================
# 8. Evaluate Accuracy
# ============================================================
correct = 0
predictions = []

for item in mini_spider:
    q = item["question"]
    gold_sql = item["answer_sql"]

    pred_raw = generate_sql(q)

    gold_norm = normalize_sql(gold_sql)
    pred_norm = normalize_sql(pred_raw)

    predictions.append((q, gold_sql, pred_raw, gold_norm, pred_norm))

    if gold_norm == pred_norm:
        correct += 1

accuracy = correct / len(mini_spider)

# ============================================================
# 9. Print Results
# ============================================================
for q, gold, pred_raw, gold_norm, pred_norm in predictions:
    print("QUESTION:", q)
    print("GOLD SQL:", gold)
    print("MODEL RAW SQL:", pred_raw)
    print("GOLD NORM:", gold_norm)
    print("PRED NORM:", pred_norm)
    print("-" * 60)

print("\nFINAL BASELINE ACCURACY (RELAXED, FIXED):", accuracy)
````

````python
# ============================================================
# 10. Additional Benchmarking Metrics for Mentor-Level Evaluation
# ============================================================
from sklearn.metrics import precision_score, recall_score, f1_score
from sentence_transformers import SentenceTransformer, util

# embedding model for similarity scoring
bench_embedder = SentenceTransformer("all-MiniLM-L6-v2")

def token_f1(gold, pred):
    """
    Token-level precision / recall / F1 benchmarking.
    Gives partial credit when SQL is close but not exact.
    """

    g = gold.split()
    p = pred.split()

    # build binary vectors over token set
    vocab = list(set(g + p))
    gold_vec = [1 if tok in g else 0 for tok in vocab]
    pred_vec = [1 if tok in p else 0 for tok in vocab]

    precision = precision_score(gold_vec, pred_vec)
    recall = recall_score(gold_vec, pred_vec)
    f1 = f1_score(gold_vec, pred_vec)

    return precision, recall, f1

def embedding_similarity(gold, pred):
    """
    Embedding-based similarity score (0–1).
    More robust than string matching.
    """
    g_emb = bench_embedder.encode(gold, convert_to_tensor=True)
    p_emb = bench_embedder.encode(pred, convert_to_tensor=True)
    sim = util.cos_sim(g_emb, p_emb).item()
    return sim

# ============================================================
# Compute Benchmarking Table
# ============================================================
benchmark_results = []

for q, gold, pred_raw, gold_norm, pred_norm in predictions:

    # token metrics
    p_token, r_token, f1_token = token_f1(gold_norm, pred_norm)

    # embedding similarity
    sim_score = embedding_similarity(gold_norm, pred_norm)

    benchmark_results.append({
        "question": q,
        "gold_sql": gold,
        "model_sql": pred_raw,
        "normalized_gold": gold_norm,
        "normalized_pred": pred_norm,
        "exact_match": int(gold_norm == pred_norm),
        "token_precision": round(p_token, 3),
        "token_recall": round(r_token, 3),
        "token_f1": round(f1_token, 3),
        "embedding_similarity": round(sim_score, 3)
    })


# ============================================================
# Display Benchmark Table
# ============================================================
print("\n================= BENCHMARK RESULTS =================\n")
for item in benchmark_results:
    print("QUESTION:", item["question"])
    print("GOLD SQL:", item["gold_sql"])
    print("MODEL SQL:", item["model_sql"])
    print("Exact Match:", item["exact_match"])
    print("Token-Level Precision:", item["token_precision"])
    print("Token-Level Recall:", item["token_recall"])
    print("Token-Level F1:", item["token_f1"])
    print("Embedding Similarity:", item["embedding_similarity"])
    print("-" * 80)

# Summary leaderboard
avg_f1 = np.mean([x["token_f1"] for x in benchmark_results])
avg_sim = np.mean([x["embedding_similarity"] for x in benchmark_results])

print("\n================= BENCHMARK SUMMARY =================")
print(f"Exact Match Accuracy      : {accuracy * 100:.2f}%")
print(f"Avg Token-Level F1 Score  : {avg_f1 * 100:.2f}%")
print(f"Avg Embedding Similarity  : {avg_sim * 100:.2f}%")
print("=====================================================\n")
````

````python
# ============================================================
# 10. EXTENDED BENCHMARK: MORE REALISTIC ACCURACY
#    - Larger eval set (10 questions)
#    - Execution accuracy using SQLite
# ============================================================

import sqlite3

# 10.1 Extended evaluation dataset (harder than mini_spider)
extended_eval = [
    {
        "question": "How many employees are older than 30?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT COUNT(*) FROM employees WHERE age > 30;"
    },
    {
        "question": "List the names of employees with salary above 50000.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name FROM employees WHERE salary > 50000;"
    },
    {
        "question": "What is the average salary of all employees?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT AVG(salary) FROM employees;"
    },
    {
        "question": "What is the minimum salary among employees younger than 30?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT MIN(salary) FROM employees WHERE age < 30;"
    },
    {
        "question": "Show the total salary paid to employees older than 40.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT SUM(salary) FROM employees WHERE age > 40;"
    },
    {
        "question": "List names of employees who earn less than 40000.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name FROM employees WHERE salary < 40000;"
    },
    {
        "question": "How many employees are between ages 25 and 35?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT COUNT(*) FROM employees WHERE age BETWEEN 25 AND 35;"
    },
    {
        "question": "List all employee names ordered by salary descending.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name FROM employees ORDER BY salary DESC;"
    },
    {
        "question": "What is the highest salary among all employees?",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT MAX(salary) FROM employees;"
    },
    {
        "question": "List names and ages of employees whose salary is between 30000 and 60000.",
        "schema": "employees(id, name, age, salary)",
        "answer_sql": "SELECT name, age FROM employees WHERE salary BETWEEN 30000 AND 60000;"
    },
]

# 10.2 Create an in-memory SQLite DB for execution accuracy
conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.execute("DROP TABLE IF EXISTS employees;")
cur.execute("""
    CREATE TABLE employees (
        id INTEGER PRIMARY KEY,
        name TEXT,
        age INTEGER,
        salary INTEGER
    );
""")

sample_rows = [
    (1, "Alice",   28, 35000),
    (2, "Bob",     45, 60000),
    (3, "Charlie", 32, 52000),
    (4, "David",   23, 30000),
    (5, "Eve",     38, 48000),
    (6, "Frank",   50, 80000),
    (7, "Grace",   27, 39000),
    (8, "Heidi",   41, 45000),
]

cur.executemany("INSERT INTO employees VALUES (?, ?, ?, ?);", sample_rows)
conn.commit()

def run_sql(sql: str):
    """
    Execute SQL and return sorted result rows.
    If SQL is invalid, return None.
    """
    try:
        cur = conn.cursor()
        cur.execute(sql)
        rows = cur.fetchall()
        # sort rows so order doesn't affect equality for our purposes,
        # except where ORDER BY is part of the query
        return rows
    except Exception as e:
        # print("SQL error:", e, "for query:", sql)
        return None

# 10.3 Evaluate over extended_eval with:
#   - string exact accuracy
#   - execution accuracy
extended_exact_correct = 0
extended_exec_correct = 0
total_extended = len(extended_eval)

extended_details = []

for item in extended_eval:
    q = item["question"]
    schema = item["schema"]
    gold_sql = item["answer_sql"]

    # Generate SQL using the SAME generate_sql() function
    pred_sql_raw = generate_sql(q)

    gold_norm = normalize_sql(gold_sql)
    pred_norm = normalize_sql(pred_sql_raw)

    # string exact accuracy (normalized)
    exact_ok = (gold_norm == pred_norm)
    if exact_ok:
        extended_exact_correct += 1

    # execution accuracy: compare outputs of gold vs predicted SQL
    gold_rows = run_sql(gold_sql.replace(";", ""))
    pred_rows = run_sql(pred_sql_raw)

    exec_ok = (gold_rows is not None) and (pred_rows is not None) and (gold_rows == pred_rows)
    if exec_ok:
        extended_exec_correct += 1

    extended_details.append({
        "question": q,
        "gold_sql": gold_sql,
        "pred_sql": pred_sql_raw,
        "exact_ok": exact_ok,
        "exec_ok": exec_ok,
        "gold_rows": gold_rows,
        "pred_rows": pred_rows,
    })

extended_exact_acc = extended_exact_correct / total_extended
extended_exec_acc = extended_exec_correct / total_extended

# 10.4 Print extended benchmark details
print("\n================= EXTENDED BENCHMARK (10 Qs) =================\n")
for d in extended_details:
    print("QUESTION:", d["question"])
    print("GOLD SQL:", d["gold_sql"])
    print("PRED SQL:", d["pred_sql"])
    print("Exact Match:", d["exact_ok"])
    print("Exec  Match:", d["exec_ok"])
    print("GOLD RESULT:", d["gold_rows"])
    print("PRED RESULT:", d["pred_rows"])
    print("-" * 80)

print("\n================= EXTENDED BENCHMARK SUMMARY =================")
print(f"String Exact Accuracy (10 Qs) : {extended_exact_acc * 100:.2f}%")
print(f"Execution Accuracy (10 Qs)    : {extended_exec_acc * 100:.2f}%")
print("================================================================\n")
````
